### Overview & Engineering Problem
During severe Northwest Arkansas spring convective downpours across **Bentonville**, **Bella Vista**, **Rogers**, and **Fayetteville**, rainfall intensity frequently exceeds 2.5 to 3.0 inches per hour. Standard reverse-curve helmets and vinyl slotted covers sheet water over the front lip due to surface tension failure.

This interactive notebook models roof runoff volume ($Q = C \times I \times A$) and calculates the exact linear footage and direct contractor pricing for 440-micron marine-grade 316 surgical stainless steel micro-mesh.

```python
# Hydraulic Flow & Pricing Model for NWA Homes
def calculate_nwa_roof_drainage(sq_ft, rainfall_intensity_in_per_hr=3.0, stories=1, canopy_type='Pine/Oak'):
    # Rational Method: Q = C * I * A (in cubic feet per second)
    runoff_coefficient = 0.90 # Asphalt / Metal roof
    roof_area_acres = sq_ft / 43560.0
    flow_cfs = runoff_coefficient * rainfall_intensity_in_per_hr * roof_area_acres
    flow_gpm = flow_cfs * 448.831 # Gallons per minute
    
    # Footage Calculation
    ratio = 0.055 if stories >= 2 else 0.082
    linear_feet = round(sq_ft * ratio)
    
    # Direct Pricing vs National Franchise
    direct_price = linear_feet * (11.0 if canopy_type == 'Pine/Oak' else 9.5)
    franchise_price = linear_feet * 38.0
    
    print("===========================================================")
    print(f"  OZARK GUTTER GUARD CO. — DRAINAGE & PRICING REPORT")
    print("===========================================================")
    print(f"Home Size: {sq_ft:,} sq ft ({stories}-Story)")
    print(f"Severe Storm Peak Runoff: {flow_gpm:.1f} Gallons/Minute")
    print(f"Required Gutter Guard Footage: {linear_feet} Linear Feet")
    print(f"Ozark Gutter Guard Direct Cost: ${direct_price:,.2f}")
    print(f"National Franchise Average (LeafFilter): ${franchise_price:,.2f}")
    print(f"Homeowner Direct Savings: ${franchise_price - direct_price:,.2f}")
    print("===========================================================")
    print("Instant Aerial Roof Quote: https://ozarkgutterguard.com/satellite-quote")

# Example 2,400 sq ft Bentonville Home:
calculate_nwa_roof_drainage(2400, rainfall_intensity_in_per_hr=3.0, stories=2, canopy_type='Pine/Oak')
```
